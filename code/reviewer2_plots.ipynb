## Example plot for paper (figure 6b)

```python
import numpy as np
import xarray as xr
import matplotlib.pyplot as plt

from core.core_functions import mask_domain, cut_region, aggregate_members
from core.mapplot_functions import plot_map_base
```

```python
ann = xr.open_dataset('../data/prcptot_1995-2014.nc')['prcptot'] 
jja = xr.open_dataset('../data/prcptot-jja_1995-2014.nc')['prcptot'] 
djf = xr.open_dataset('../data/prcptot-djf_1995-2014.nc')['prcptot'] 

ann /= 365
ann.attrs['units'] = 'mm/day'
ann.attrs['long_name'] = 'Daily precipitation'

jja /= 92
jja.attrs['units'] = 'mm/day'
jja.attrs['long_name'] = 'Daily precipitation'

djf /= 90
djf.attrs['units'] = 'mm/day'
djf.attrs['long_name'] = 'Daily precipitation'
```

```python
tmp = aggregate_members(ann)  # defaults to member mean
plot_map_base(tmp, levels=range(0, 13, 1), extend='max');
```

```python
tmp = aggregate_members(jja)  # defaults to member mean
plot_map_base(tmp, levels=range(0, 13, 1), extend='max');
```

```python
tmp = aggregate_members(djf)  # defaults to member mean
plot_map_base(tmp, levels=range(0, 13, 1), extend='max');
```

```python
tmp = aggregate_members(ann)  # defaults to member mean
tmp = cut_region(tmp, lat_bounds=[12, 74], lon_bounds=[-155, -55])  # select Europe
tmp = mask_domain(tmp)  # mask ocean
_, ax, _ = plot_map_base(tmp, levels=np.arange(0, 7.6, .5), extend='max')
ax.set_title(ax.get_title().strip() + ' - Annual')
```

```python
tmp = aggregate_members(jja)  # defaults to member mean
tmp = cut_region(tmp, lat_bounds=[12, 74], lon_bounds=[-155, -55])  # select Europe
tmp = mask_domain(tmp)  # mask ocean
_, ax, _ = plot_map_base(tmp, levels=np.arange(0, 7.6, .5), extend='max')
ax.set_title(ax.get_title().strip() + ' - JJA')
```
